```python
import csv
import json
```

```python
purchases = {}
```

```python
with open("purchase_log.txt", encoding="utf-8") as f:
    next(f)
    for line in f:
        purchase = json.loads(line)
        purchases[purchase["user_id"]] = purchase["category"]
```

```python
list(purchases.items())[:2]
```

```python
with open("visit_log.csv", encoding="utf-8") as visits, open("funnel.csv", "w", encoding="utf-8", newline="") as funnel:

    reader = csv.reader(visits)
    writer = csv.writer(funnel)

    header = next(reader)
    writer.writerow(header + ["category"])

    for row in reader:
        user_id = row[0]

        if user_id in purchases:
            writer.writerow(row + [purchases[user_id]])
```
